```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import seaborn as sns
visa_df=pd.read_csv(r"C:\Users\suhan\OneDrive\Documents\Naresh_IT\Data Files\Visadataset.csv")
```

```python
cat = []
num = []

for col in visa_df.columns:
    if visa_df[col].dtype == 'object':
        cat.append(col)
    else:
        num.append(col)
cat
```

```python
for i in cat[1:]:
    keys=visa_df[i].value_counts().keys()
    values=visa_df[i].value_counts().values
    plt.figure(figsize=(10,4))
    plt.bar(keys,values)
    plt.xlabel('labels')
    plt.ylabel('count')
    plt.title(f'{i}_barchart')
    plt.savefig(f'{i}_bar.jpg')
    plt.show()
```

```python
for i in cat[1:]:
    keys=visa_df[i].value_counts().keys()
    values=visa_df[i].value_counts().values
    plt.figure(figsize=(10,4))
    plt.pie(values,labels=keys)
    plt.title(f'{i}_barchart')
    plt.savefig(f'{i}_bar.jpg')
    plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
hh=visa_df['continent'].unique()
count=[]
for i in hh:
    c=visa_df['continent']==i
    count.append(np.sum(c))
plt.pie(count,labels=hh)
plt.show()
    
```
